#09번 문제

삽입된 노드의 부모와 삼촌 노드가 모두 레드일 때 발생한다. 부모와 삼촌을 블랙으로 바꾸고 조상을 레드로 바꾼다. 이때 조상의 부모가 레드라면 다시 동일한 문제가 반복된다. 동일한 문제가 반복된다면 각 노드에 아래와 같은 영향을 미치게 된다. 
p1과 p2에 미치는 영향:
    새롭게 삽입한 노드인 p가 레드이면 부모(p1)이 이미 레드이기 때문에 부모 노드(p1)과 p1의 형제노드 중 레드인 노드를 블랙으로 바꾸고 조상 노드(p2)를 레드로 바꾼다.
p3과 p4에 미치는 영향:
    p2가 레드로 변했는데 p3도 레드라면 p3과 p3의 형제 중 레드인 노드를 블랙으로 바꾸고 p4를 레드로 바꾼다.
p5와 p6에 미치는 영향:
    p4가 레드로 바뀌었을 때 p5가 레드라면 이 과정이 다시 반복된다. p5의 형제 노드가 레드라면 p5와 형제를 블랙으로, p6을 레드로 바꾸게 된다.

#14번 문제

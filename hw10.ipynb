#16번 문제
수도코드: 
topologicalSort(Graph G)
    // 1. 모든 노드의 진입 차수(in-degree) 계산
    for each vertex v in G:
        in_degree[v] = 0
    for each vertex v in G:
        for each neighbor u of v:
            in_degree[u] = in_degree[u] + 1

    // 2. 진입 차수가 0인 노드를 스택에 삽입
    Create an empty stack S
    for each vertex v in G:
        if in_degree[v] == 0:
            S.push(v)

    // 3. 위상 정렬 수행
    Create an empty list result
    while S is not empty:
        v = S.pop()
        result.append(v)  // 정렬 결과 리스트에 추가
        
        // 연결된 이웃 노드들의 진입 차수 감소 및 조건 확인
        for each neighbor u of v:
            in_degree[u] = in_degree[u] - 1
            if in_degree[u] == 0:
                S.push(u)

    // 4. 결과 반환 (사이클이 없다면 정렬 성공)
    if length(result) < number of vertices in G:
        return "Graph has a cycle (Topological sort impossible)"
    else:
        return result

가능한 위상 정렬 순서의 개수: 20가지

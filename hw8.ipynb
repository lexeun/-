#01번 문제

첫 번째 방법은 체이닝이고, 두 번째 방법은 open addressing(개방 주소 방법)이다. 체이닝은 같은 자리에서 충돌이 일어난 원소들은 하나씩의 연결리스트로 관리하고, 개방 주소 방법은 주어진 배열 안에서 해결한다.

체이닝은 해시함수가 데이터를 특정 버킷에 배정하기만 하면 되므로 로직이 단순하다는 장점이 있지만 데이터들이 메모리 곳곳에 흩어진 채 저장될 가능성이 높아 cpu 캐시 적중률이 떨어질 수 있다는 단점이 있다. 개방 주소 방법은 별도의 연결 리스트 없이 테이블 내의 빈 공간을 활용하기 때문에 추가 메모리가 필요 없다는 장점이 있지만, 중간 데이터를 삭제하면 검색 경로가 끊길 수 있어 표시를 남기는 등 별도의 관리가 필요하다는 단점이 있다. 

#02-1

첫번째는 테이블 공간이 낭비된다는 점이다. 해시 테이블의 크기와 해시 함수가 생성하는 값의 범위가 다르기 때문에 어떤 키가 들어오덜도 절대 사용되지 않는 유휴 공간이 생기게 된다. 두번째는 해시 함수에서 나누는 수로 2^n을 사용하면, 입력 데이터의 하위비트만 해시 값 결정에 영향을 준다는 점이다. 이 경우에 입력 데이터에 일정한 패턴이 있다면, 특정 버킷에만 데이터가 몰리는 충돌이 빈번하게 발생히게 된다.

#02-2

결정성을 만족하지 않기 때문이다. 해시 함수는 동일한 입력값에 대해 항상 동일한 해시 값을 반환해야 하는데 위의 식처럼 계산 과정에 random()함수가 포함되면 데이터를 저장할 때의 주소와 나중에 호출할 때의 주소가 달라진다. 즉 데이터를 저장해 놓고도 어디에 있는지 찾을 수 없는 상태가 되기 때문에 해시 함수가 될 수 없다.

#02-7

리스트의 맨 앞이 아닌 맨 뒤에 삽입한다면 효율성이 낮아진다. 맨 앞에 삽입하는경우에는 헤드 포인터만 수정하면 되기 때문에 O(1)의 시간이 걸리지만 맨 뒤에 삽입하려면 처음주터 끝까지 삽입해야하기 때문에 O(n)의 시간이 걸리게 된다.

#02-8

삽입: 저하된다. 아무 위치에나 넣을 수 없고 정렬된 순서에 맞는 위치를 찾기 위해 리스트를 탐색해야 하기 때문에 평균적으로 더 많은 비교가 필요하다.

삭제: 유지되거나 약간 향상된다. 삭제 대항을 찾기 위해 탐색해야 하는 과정은 동일하지만 찾는 값보다 큰 값이 나오면 리스트 끝까지 가지 않아도 해당 값이 없음을 알 수 있기 때문에 일찍 종료할 수 있다.

성공적인 검색: 유지된다. 찾고자 하는 값이 리스트에 있는 경유 정렬 여부와 관계없이 해당 노드까지 도달해야 하므로 효율성 변화가 거의 없다.

실패하는 검색: 향상된다. 정렬되지 않은 리스트는 끝까지 가봐야 없다는 것을 알 수 있지만, 정렬된 리스트는 자신보다 큰 값을 만나는 즉시 검색을 중단할 수 있다. 따라서 평군적으로 탐색하는 노드 수가 절반으로 줄어들게 된다.

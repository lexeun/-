#03번 문제
1) 5->3->1->8->7->6->9
2) 1->3->5->6->7->8->9
3) 1->3->6->7->9->8->5
4) 이진 검색트리는 항상 왼쪽 서브트리의 모든 값은 루트노드보다 작거나 같고, 오른쪽 서브트리의 모든 값은 루트노드보다 크거나 같다. 왼쪽 서브트리를 Tl 루트 노드를 r, 오른쪽 서브트리를 Tr, 트리의 높이를 h라고 해보자. h가 1일 때는 단 한 개의 노드를 갖기 때문에 당연히 정렬된 순서이다. 귀납법을 이용해서 h보다 작은 모든 k에 대해서 이진 검색트리의 중위 순회가 노드들을 정렬된 순서로 방문한다고 가정하자. 이진 검색트리에서 중위순회는 Tl, r, Tr순으로 방문한다. 앞서 말한 것처럼 Tl의 모든 키는 r의 키보다 작고, r의 키는 Tr의 모든 키보다 작기 때문에 중위순회 Tl->r->Tr은 정렬된 순서이다.
